```python
# =========================================================
# PREPROCESSING DATASET TMDB 5000 MOVIES
# CONTENT-BASED FILTERING
# =========================================================
#
# METODE YANG DIGUNAKAN:
#
# 1. Data Cleaning
#    - Remove Duplicate
#    - Missing Value Handling
#
# 2. Text Preprocessing
#    - Case Folding
#    - Remove Special Character
#    - Remove Extra Space
#
# 3. Feature Extraction
#    - Genre Extraction
#    - Keyword Extraction
#
# 4. Feature Engineering
#    - Combined Features
#
# =========================================================

# =========================================================
# 1. IMPORT LIBRARY
# =========================================================

import pandas as pd
import ast
import re

# =========================================================
# 2. LOAD DATASET ASLI
# =========================================================
# PASTIKAN FILE YANG DIGUNAKAN:
# tmdb_5000_movies.csv
# BUKAN:
# tmdb_5000_movies_cleaned.csv

df = pd.read_csv('/content/drive/MyDrive/New Dataset/tmdb_5000_movies.csv')

# =========================================================
# 3. INFORMASI DATASET
# =========================================================

print("=================================================")
print("INFORMASI DATASET")
print("=================================================")

print("\nJumlah Data Awal :", len(df))

print("\nDaftar Kolom Dataset:\n")
print(df.columns)

# =========================================================
# 4. HAPUS DUPLIKAT
# METODE : Data Cleaning
# =========================================================

df = df.drop_duplicates()

print("\nJumlah Data Setelah Hapus Duplikat :", len(df))

# =========================================================
# 5. MENANGANI MISSING VALUE
# METODE : Missing Value Handling
# =========================================================

text_columns = [
    'genres',
    'keywords',
    'overview',
    'tagline',
    'homepage'
]

# Isi data kosong dengan string kosong
for col in text_columns:

    # cek apakah kolom tersedia
    if col in df.columns:

        df[col] = df[col].fillna('')

# Hapus data yang tidak memiliki title
df = df.dropna(subset=['title'])

print("\nMissing Value berhasil ditangani")

# =========================================================
# 6. KONVERSI TANGGAL
# METODE : Data Transformation
# =========================================================

df['release_date'] = pd.to_datetime(
    df['release_date'],
    errors='coerce'
)

# Ambil tahun rilis
df['release_year'] = df['release_date'].dt.year

print("\nKonversi release_date berhasil")

# =========================================================
# 7. TEXT PREPROCESSING
# METODE:
# - Case Folding
# - Remove Punctuation
# - Remove Extra Space
# =========================================================

def clean_text(text):

    # Ubah menjadi huruf kecil
    text = str(text).lower()

    # Hapus karakter selain huruf dan angka
    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)

    # Hapus spasi berlebih
    text = re.sub(r'\s+', ' ', text).strip()

    return text

# Bersihkan overview
df['overview_clean'] = df['overview'].apply(clean_text)

print("\nText preprocessing berhasil")

# =========================================================
# 8. EKSTRAK GENRE DAN KEYWORDS
# METODE : Feature Extraction
# =========================================================

def extract_names(obj):

    # Jika kosong
    if pd.isna(obj):
        return ''

    try:

        # Konversi string JSON ke Python list
        data = ast.literal_eval(obj)

        # Jika format list
        if isinstance(data, list):

            names = []

            for item in data:

                # Ambil value name
                if isinstance(item, dict):

                    if 'name' in item:
                        names.append(item['name'])

            return ' '.join(names)

        return ''

    except:
        return ''

# =========================================================
# EKSTRAK GENRES
# =========================================================

if 'genres' in df.columns:

    df['genres_clean'] = df['genres'].apply(
        extract_names
    )

else:

    df['genres_clean'] = ''

# =========================================================
# EKSTRAK KEYWORDS
# =========================================================

if 'keywords' in df.columns:

    df['keywords_clean'] = df['keywords'].apply(
        extract_names
    )

else:

    df['keywords_clean'] = ''

# Isi NaN jika masih ada
df['genres_clean'] = df['genres_clean'].fillna('')
df['keywords_clean'] = df['keywords_clean'].fillna('')

print("\nEkstraksi Genre dan Keyword berhasil")

# =========================================================
# 9. FEATURE ENGINEERING
# METODE : Combined Features
# =========================================================
# Genre dibuat 2x agar bobot genre lebih tinggi

df['combined_features'] = (
    df['genres_clean'] + ' ' +
    df['genres_clean'] + ' ' +
    df['keywords_clean'] + ' ' +
    df['overview_clean']
)

print("\nCombined Features berhasil dibuat")

# =========================================================
# 10. PILIH KOLOM YANG DIGUNAKAN
# =========================================================

final_columns = [
    'id',
    'title',
    'genres_clean',
    'keywords_clean',
    'overview_clean',
    'combined_features',
    'popularity',
    'vote_average',
    'vote_count',
    'release_year'
]

final_df = df[final_columns]

# =========================================================
# 11. SIMPAN HASIL PREPROCESSING
# =========================================================

final_df.to_csv(
    'tmdb_5000_movies_cleaned.csv',
    index=False
)

# =========================================================
# 12. HASIL PREPROCESSING
# =========================================================

print("\n=================================================")
print("HASIL PREPROCESSING")
print("=================================================")

print("\nJumlah Data Akhir :", len(final_df))

print("\nJumlah Kolom Akhir :", len(final_df.columns))

print("\nContoh Hasil Preprocessing:\n")

print(
    final_df[
        [
            'title',
            'genres_clean',
            'keywords_clean'
        ]
    ].head()
)

print("\n=================================================")
print("DATASET BERHASIL DISIMPAN")
print("Nama File : tmdb_5000_movies_cleaned.csv")
print("=================================================")
```

```python
# ============================================================
# SISTEM REKOMENDASI FILM
# TF-IDF BERBOBOT + COSINE SIMILARITY
# Evaluasi: Precision@10 dan NDCG@10 dalam persen
# Bisa input judul film sendiri
# Similarity rekomendasi minimal 0.5
# Evaluasi dibuat lebih realistis agar tidak 100%
# ============================================================

import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.preprocessing import normalize

from scipy.sparse import hstack


# ============================================================
# 1. LOAD DATASET
# ============================================================

df = pd.read_csv("tmdb_5000_movies_cleaned.csv")

print("Jumlah data awal:", df.shape)
print("Kolom dataset:")
print(df.columns)


# ============================================================
# 2. PREPROCESSING DATA
# ============================================================

df["id"] = df["id"].fillna("")
df["title"] = df["title"].fillna("")
df["genres_clean"] = df["genres_clean"].fillna("")
df["keywords_clean"] = df["keywords_clean"].fillna("")
df["overview_clean"] = df["overview_clean"].fillna("")
df["combined_features"] = df["combined_features"].fillna("")
df["popularity"] = df["popularity"].fillna(0)
df["vote_average"] = df["vote_average"].fillna(0)
df["vote_count"] = df["vote_count"].fillna(0)
df["release_year"] = df["release_year"].fillna(0)

# Hapus data tanpa judul
df = df[df["title"] != ""].reset_index(drop=True)

print("\nJumlah data setelah preprocessing:", df.shape)


# ============================================================
# 3. TF-IDF UNTUK SETIAP FITUR
# ============================================================

# Genre dibuat sendiri karena genre sangat penting dalam film
tfidf_genres = TfidfVectorizer(
    stop_words="english"
)

# Keyword juga penting, tetapi bobotnya di bawah genre
tfidf_keywords = TfidfVectorizer(
    stop_words="english",
    max_features=3000
)

# Overview biasanya panjang, jadi bobotnya lebih kecil
tfidf_overview = TfidfVectorizer(
    stop_words="english",
    max_features=5000,
    ngram_range=(1, 2),
    min_df=2
)

genre_matrix = tfidf_genres.fit_transform(df["genres_clean"])
keyword_matrix = tfidf_keywords.fit_transform(df["keywords_clean"])
overview_matrix = tfidf_overview.fit_transform(df["overview_clean"])

print("\nUkuran TF-IDF Genre:", genre_matrix.shape)
print("Ukuran TF-IDF Keyword:", keyword_matrix.shape)
print("Ukuran TF-IDF Overview:", overview_matrix.shape)


# ============================================================
# 4. PEMBOBOTAN FITUR
# ============================================================

# Bobot ini bukan manipulasi skor akhir.
# Bobot diberikan sebelum cosine similarity dihitung.
# Jadi similarity tetap valid secara metodologi.

GENRE_WEIGHT = 5.0
KEYWORD_WEIGHT = 3.0
OVERVIEW_WEIGHT = 1.0

weighted_tfidf_matrix = hstack([
    genre_matrix * GENRE_WEIGHT,
    keyword_matrix * KEYWORD_WEIGHT,
    overview_matrix * OVERVIEW_WEIGHT
])

# Normalisasi agar cocok untuk cosine similarity
weighted_tfidf_matrix = normalize(
    weighted_tfidf_matrix,
    norm="l2",
    axis=1
)

print("\nUkuran Weighted TF-IDF Matrix:", weighted_tfidf_matrix.shape)


# ============================================================
# 5. COSINE SIMILARITY
# ============================================================

cosine_sim = cosine_similarity(
    weighted_tfidf_matrix,
    weighted_tfidf_matrix
)

print("Ukuran Cosine Similarity Matrix:", cosine_sim.shape)


# ============================================================
# 6. MEMBUAT INDEX BERDASARKAN JUDUL FILM
# ============================================================

indices = pd.Series(
    df.index,
    index=df["title"].str.lower()
).drop_duplicates()


# ============================================================
# 7. FUNGSI REKOMENDASI FILM
# ============================================================

def recommend_movies(title, top_n=10, min_similarity=0.5):
    """
    Memberikan rekomendasi film berdasarkan judul film
    menggunakan TF-IDF berbobot dan Cosine Similarity.

    Parameter:
    title          : judul film input
    top_n          : jumlah rekomendasi
    min_similarity : batas minimal cosine similarity

    Return:
    DataFrame rekomendasi film
    """

    title_lower = title.lower().strip()

    if title_lower not in indices:
        return None

    idx = indices[title_lower]

    # Ambil skor similarity film input terhadap semua film
    sim_scores = list(enumerate(cosine_sim[idx]))

    # Urutkan berdasarkan similarity tertinggi
    sim_scores = sorted(
        sim_scores,
        key=lambda x: x[1],
        reverse=True
    )

    # Hilangkan film itu sendiri
    # Lalu ambil hanya film dengan similarity >= 0.5
    sim_scores = [
        score for score in sim_scores
        if score[0] != idx and score[1] >= min_similarity
    ]

    # Ambil Top-N
    sim_scores = sim_scores[:top_n]

    # Jika tidak ada film yang memenuhi batas similarity
    if len(sim_scores) == 0:
        return pd.DataFrame()

    movie_indices = [i[0] for i in sim_scores]
    similarity_scores = [i[1] for i in sim_scores]

    recommendations = df.loc[
        movie_indices,
        [
            "id",
            "title",
            "genres_clean",
            "release_year",
            "vote_average",
            "vote_count",
            "popularity"
        ]
    ].copy()

    recommendations["similarity_score"] = similarity_scores
    recommendations["similarity_score (%)"] = recommendations["similarity_score"] * 100

    return recommendations.reset_index(drop=True)


# ============================================================
# 8. FUNGSI PENCARIAN FILM
# ============================================================

def search_movie(keyword):
    """
    Mencari film berdasarkan keyword judul.
    Berguna jika judul yang dimasukkan tidak ditemukan secara persis.
    """

    keyword = keyword.lower().strip()

    result = df[
        df["title"].str.lower().str.contains(keyword, na=False, regex=False)
    ]

    return result[
        [
            "id",
            "title",
            "genres_clean",
            "release_year",
            "vote_average",
            "popularity"
        ]
    ].reset_index(drop=True)


# ============================================================
# 9. FUNGSI PARSING GENRE DAN KEYWORD UNTUK EVALUASI
# ============================================================

KNOWN_GENRES = [
    "action",
    "adventure",
    "animation",
    "comedy",
    "crime",
    "documentary",
    "drama",
    "family",
    "fantasy",
    "history",
    "horror",
    "music",
    "mystery",
    "romance",
    "science fiction",
    "tv movie",
    "thriller",
    "war",
    "western"
]


def parse_genres(genre_text):
    """
    Mengubah teks genre menjadi set genre yang lebih rapi.

    Contoh:
    'Action Adventure Science Fiction'
    menjadi:
    {'action', 'adventure', 'science fiction'}
    """

    if not isinstance(genre_text, str):
        return set()

    genre_text = genre_text.lower()

    genres = set()

    for genre in KNOWN_GENRES:
        if genre in genre_text:
            genres.add(genre)

    return genres


def text_to_set(text):
    """
    Mengubah keyword menjadi set kata.
    """

    if not isinstance(text, str):
        return set()

    text = text.lower().strip()

    if text == "":
        return set()

    return set(text.split())


# ============================================================
# 10. EVALUASI PRECISION@10 DAN NDCG@10
# ============================================================

def evaluate_precision_ndcg_at_10():
    """
    Evaluasi sistem rekomendasi menggunakan Precision@10 dan NDCG@10.

    Evaluasi dibuat lebih realistis agar tidak 100%.

    Film dianggap relevan jika:
    1. Memiliki minimal 2 genre yang sama, atau
    2. Memiliki minimal 1 genre yang sama dan minimal 2 keyword yang sama.

    Precision@10:
    Mengukur berapa banyak film relevan dari 10 rekomendasi teratas.

    NDCG@10:
    Mengukur kualitas urutan rekomendasi.
    Film relevan di posisi atas mendapat nilai lebih besar.
    """

    k = 10

    genre_sets = df["genres_clean"].apply(parse_genres).tolist()
    keyword_sets = df["keywords_clean"].apply(text_to_set).tolist()

    precision_scores = []
    ndcg_scores = []

    jumlah_film = len(df)

    for idx in range(jumlah_film):
        target_genres = genre_sets[idx]
        target_keywords = keyword_sets[idx]

        # Lewati film yang tidak punya genre
        if len(target_genres) == 0:
            continue

        # Ambil skor cosine similarity
        scores = cosine_sim[idx].copy()

        # Supaya film itu sendiri tidak ikut direkomendasikan
        scores[idx] = -1

        # Ambil Top-10 berdasarkan similarity tertinggi
        top_10_indices = np.argsort(scores)[::-1][:k]

        relevances = []

        for movie_idx in top_10_indices:
            candidate_genres = genre_sets[movie_idx]
            candidate_keywords = keyword_sets[movie_idx]

            same_genres = len(target_genres.intersection(candidate_genres))
            same_keywords = len(target_keywords.intersection(candidate_keywords))

            # Aturan relevansi lebih ketat
            if same_genres >= 2 or (same_genres >= 1 and same_keywords >= 2):
                relevances.append(1)
            else:
                relevances.append(0)

        # ====================================================
        # PRECISION@10
        # ====================================================

        hits = sum(relevances)
        precision = hits / k
        precision_scores.append(precision)

        # ====================================================
        # NDCG@10
        # ====================================================

        dcg = 0

        for i, rel in enumerate(relevances):
            rank = i + 1
            dcg += rel / np.log2(rank + 1)

        # Hitung semua film relevan sebagai pembanding ideal
        all_relevances = []

        for j in range(jumlah_film):
            if j == idx:
                continue

            candidate_genres = genre_sets[j]
            candidate_keywords = keyword_sets[j]

            same_genres = len(target_genres.intersection(candidate_genres))
            same_keywords = len(target_keywords.intersection(candidate_keywords))

            if same_genres >= 2 or (same_genres >= 1 and same_keywords >= 2):
                all_relevances.append(1)
            else:
                all_relevances.append(0)

        ideal_relevances = sorted(all_relevances, reverse=True)[:k]

        idcg = 0

        for i, rel in enumerate(ideal_relevances):
            rank = i + 1
            idcg += rel / np.log2(rank + 1)

        if idcg > 0:
            ndcg = dcg / idcg
            ndcg_scores.append(ndcg)

    precision_mean = np.mean(precision_scores)
    ndcg_mean = np.mean(ndcg_scores)

    evaluation_result = {
        "K": k,
        "Jumlah film dievaluasi": len(precision_scores),

        "Precision@10": precision_mean,
        "NDCG@10": ndcg_mean,

        "Precision@10 (%)": precision_mean * 100,
        "NDCG@10 (%)": ndcg_mean * 100
    }

    return evaluation_result


# ============================================================
# 11. INPUT JUDUL FILM DARI USER
# ============================================================

print("\n================================================")
print("SISTEM REKOMENDASI FILM")
print("================================================")

judul_input = input("Masukkan judul film: ")

hasil_rekomendasi = recommend_movies(
    title=judul_input,
    top_n=10,
    min_similarity=0.5
)

if hasil_rekomendasi is not None:

    if len(hasil_rekomendasi) > 0:
        print(f"\nRekomendasi film yang mirip dengan '{judul_input}':")
        print("Hanya menampilkan film dengan cosine similarity minimal 0.5")
        print(hasil_rekomendasi)

    else:
        print(f"\nFilm '{judul_input}' ditemukan, tetapi tidak ada rekomendasi dengan similarity >= 0.5.")
        print("Coba turunkan min_similarity menjadi 0.4 jika ingin rekomendasi lebih banyak.")

else:
    print(f"\nFilm '{judul_input}' tidak ditemukan dalam dataset.")
    print("Coba cari judul yang mirip.")

    hasil_pencarian = search_movie(judul_input)

    if len(hasil_pencarian) > 0:
        print("\nMungkin maksud Anda salah satu film berikut:")
        print(hasil_pencarian.head(10))
    else:
        print("\nTidak ada judul film yang mirip ditemukan.")


# ============================================================
# 12. EVALUASI SISTEM REKOMENDASI
# ============================================================

print("\n================================================")
print("HASIL EVALUASI SISTEM")
print("================================================")

evaluation = evaluate_precision_ndcg_at_10()

print("K:", evaluation["K"])
print("Jumlah film dievaluasi:", evaluation["Jumlah film dievaluasi"])

print("Precision@10:", evaluation["Precision@10"])
print("NDCG@10:", evaluation["NDCG@10"])

print("Precision@10 (%):", f"{evaluation['Precision@10 (%)']:.2f}%")
print("NDCG@10 (%):", f"{evaluation['NDCG@10 (%)']:.2f}%")
```

```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import normalize
from scipy.sparse import hstack

# ============================================================
# 1. TF-IDF SETIAP FITUR
# ============================================================

tfidf_genres = TfidfVectorizer(stop_words="english")
genre_matrix = tfidf_genres.fit_transform(df["genres_clean"])

tfidf_keywords = TfidfVectorizer(
    stop_words="english",
    max_features=3000
)
keyword_matrix = tfidf_keywords.fit_transform(df["keywords_clean"])

tfidf_overview = TfidfVectorizer(
    stop_words="english",
    max_features=5000,
    ngram_range=(1, 2),
    min_df=2
)
overview_matrix = tfidf_overview.fit_transform(df["overview_clean"])

# ============================================================
# 2. WEIGHTED TF-IDF (SESUAI KODE KAMU)
# ============================================================

GENRE_WEIGHT = 5.0
KEYWORD_WEIGHT = 3.0
OVERVIEW_WEIGHT = 1.0

weighted_tfidf_matrix = hstack([
    genre_matrix * GENRE_WEIGHT,
    keyword_matrix * KEYWORD_WEIGHT,
    overview_matrix * OVERVIEW_WEIGHT
])

# ============================================================
# 3. NORMALISASI (WAJIB SESUAI KODE ASLI)
# ============================================================

weighted_tfidf_matrix = normalize(
    weighted_tfidf_matrix,
    norm="l2",
    axis=1
)

# ============================================================
# 4. MENAMPILKAN VEKTOR TF-IDF FINAL
# ============================================================

def show_tfidf_vector(title, top_n=20):
    # cari index film
    idx = df[df["title"] == title].index[0]

    # ambil vektor hasil akhir
    vector = weighted_tfidf_matrix[idx].toarray().flatten()

    # ambil semua term gabungan (tidak dipisah karena sudah hstack)
    feature_names = (
        list(tfidf_genres.get_feature_names_out()) +
        list(tfidf_keywords.get_feature_names_out()) +
        list(tfidf_overview.get_feature_names_out())
    )

    # buat dataframe
    result = pd.DataFrame({
        "Term": feature_names,
        "TF-IDF Value": vector
    })

    # hanya yang > 0
    result = result[result["TF-IDF Value"] > 0]

    # sorting terbesar
    result = result.sort_values(by="TF-IDF Value", ascending=False)

    print(f"\n=== VEKTOR TF-IDF FINAL: {title} ===")
    return result.head(top_n)

# ============================================================
# 5. CONTOH PAKAI
# ============================================================

show_tfidf_vector("Avatar", top_n=20)
```

```python
# ============================================================
# 1. IMPORT LIBRARY
# ============================================================
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import normalize
from sklearn.metrics.pairwise import cosine_similarity
from scipy.sparse import hstack

# ============================================================
# 2. TF-IDF (3 FITUR)
# ============================================================

# GENRE
tfidf_genres = TfidfVectorizer(stop_words="english")
genre_matrix = tfidf_genres.fit_transform(df["genres_clean"])

# KEYWORD
tfidf_keywords = TfidfVectorizer(
    stop_words="english",
    max_features=3000
)
keyword_matrix = tfidf_keywords.fit_transform(df["keywords_clean"])

# OVERVIEW
tfidf_overview = TfidfVectorizer(
    stop_words="english",
    max_features=5000,
    ngram_range=(1,2),
    min_df=2
)
overview_matrix = tfidf_overview.fit_transform(df["overview_clean"])

# ============================================================
# 3. WEIGHTED TF-IDF
# ============================================================

GENRE_WEIGHT = 5.0
KEYWORD_WEIGHT = 3.0
OVERVIEW_WEIGHT = 1.0

weighted_tfidf_matrix = hstack([
    genre_matrix * GENRE_WEIGHT,
    keyword_matrix * KEYWORD_WEIGHT,
    overview_matrix * OVERVIEW_WEIGHT
])

# ============================================================
# 4. NORMALISASI
# ============================================================

weighted_tfidf_matrix = normalize(
    weighted_tfidf_matrix,
    norm="l2",
    axis=1
)

# ============================================================
# 5. COSINE SIMILARITY MATRIX
# ============================================================

cosine_sim = cosine_similarity(weighted_tfidf_matrix)

print("Shape Cosine Similarity:", cosine_sim.shape)

# ============================================================
# 6. AMBIL VEKTOR COSINE SIMILARITY (1 FILM)
# ============================================================

def get_similarity_vector(title):
    idx = df[df["title"] == title].index[0]
    return cosine_sim[idx]

# ============================================================
# 7. REKOMENDASI FILM
# ============================================================

def recommend_movies(title, top_n=10):
    idx = df[df["title"] == title].index[0]

    sim_scores = list(enumerate(cosine_sim[idx]))

    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    sim_scores = sim_scores[1:top_n+1]  # skip dirinya sendiri

    movie_indices = [i[0] for i in sim_scores]

    result = df.iloc[movie_indices][["title"]].copy()
    result["cosine_similarity"] = [i[1] for i in sim_scores]

    return result

# ============================================================
# 8. CONTOH OUTPUT
# ============================================================

# VEKTOR COSINE SIMILARITY
vector = get_similarity_vector("Avatar")
print(vector)

# TOP 10 REKOMENDASI
recommend_movies("Avatar", 10)
```
